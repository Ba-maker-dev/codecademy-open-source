```python
"""
Projet: Maintenance Prédictive Aéronautique par IA
Dataset: NASA C-MAPSS (Commercial Modular Aero-Propulsion System Simulation)
Objectif: Prédire la Durée de Vie Restante (RUL) des turboréacteurs
"""

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import warnings
warnings.filterwarnings('ignore')

# Configuration de l'affichage
plt.style.use('seaborn-v0_8-darkgrid')
sns.set_palette("husl")

print("="*70)
print("MAINTENANCE PRÉDICTIVE AÉRONAUTIQUE - NASA C-MAPSS")
print("="*70)

# ============================================================================
# PHASE 1: CHARGEMENT ET ANALYSE EXPLORATOIRE DES DONNÉES (EDA)
# ============================================================================

print("\n[PHASE 1] Chargement et Analyse Exploratoire des Données")
print("-"*70)

# Définition des noms de colonnes
column_names = ['unit_nr', 'time_cycles'] + \
               [f'op_setting_{i}' for i in range(1, 4)] + \
               [f'sensor_{i}' for i in range(1, 22)]

# Chargement des données d'entraînement
# OPTION A: Si vous utilisez Google Colab, décommentez ces lignes:
# from google.colab import files
# uploaded = files.upload()  # Upload train_FD001.txt
# uploaded = files.upload()  # Upload test_FD001.txt
# uploaded = files.upload()  # Upload RUL_FD001.txt

# OPTION B: Si les fichiers sont dans le même dossier:
try:
    train_df = pd.read_csv('train_FD001.txt', sep=r'\s+', header=None, names=column_names)
    test_df = pd.read_csv('test_FD001.txt', sep=r'\s+', header=None, names=column_names)
    rul_true = pd.read_csv('RUL_FD001.txt', sep=r'\s+', header=None, names=['RUL'])

    print(f"✓ Données chargées avec succès")
    print(f"  - Train: {train_df.shape[0]} lignes, {train_df.shape[1]} colonnes")
    print(f"  - Test: {test_df.shape[0]} lignes, {test_df.shape[1]} colonnes")
    print(f"  - Nombre de moteurs (train): {train_df['unit_nr'].nunique()}")
    print(f"  - Nombre de moteurs (test): {test_df['unit_nr'].nunique()}")

except FileNotFoundError:
    print("⚠ ATTENTION: Fichiers de données non trouvés!")
    print("Ce code est un template. Placez les fichiers suivants dans le répertoire:")
    print("  - train_FD001.txt")
    print("  - test_FD001.txt")
    print("  - RUL_FD001.txt")
    print("\nCréation de données simulées pour démonstration...")

    # Création de données simulées pour test du code
    np.random.seed(42)
    n_engines = 10
    train_data = []
    for engine in range(1, n_engines + 1):
        max_cycles = np.random.randint(150, 250)
        for cycle in range(1, max_cycles + 1):
            row = [engine, cycle] + list(np.random.randn(24) * 10 + 50)
            train_data.append(row)

    train_df = pd.DataFrame(train_data, columns=column_names)
    test_df = train_df.groupby('unit_nr').head(100).copy()
    rul_true = pd.DataFrame({'RUL': np.random.randint(50, 150, n_engines)})

# Affichage des premières lignes
print("\n📊 Aperçu des données d'entraînement:")
print(train_df.head())

# Statistiques descriptives
print("\n📈 Statistiques descriptives:")
print(train_df.describe().T)

# ============================================================================
# PHASE 2: PRÉTRAITEMENT - CALCUL DU RUL
# ============================================================================

print("\n[PHASE 2] Prétraitement et Calcul du RUL")
print("-"*70)

# Calcul du RUL pour les données d'entraînement
# RUL = Cycle Maximum du moteur - Cycle Actuel
train_df['RUL'] = train_df.groupby('unit_nr')['time_cycles'].transform('max') - train_df['time_cycles']

print(f"✓ Colonne RUL créée")
print(f"  - RUL moyen: {train_df['RUL'].mean():.2f} cycles")
print(f"  - RUL max: {train_df['RUL'].max()} cycles")
print(f"  - RUL min: {train_df['RUL'].min()} cycles")

# Visualisation de l'évolution du RUL pour quelques moteurs
plt.figure(figsize=(14, 6))
for engine_id in train_df['unit_nr'].unique()[:5]:
    engine_data = train_df[train_df['unit_nr'] == engine_id]
    plt.plot(engine_data['time_cycles'], engine_data['RUL'], label=f'Moteur {engine_id}', linewidth=2)

plt.xlabel('Cycles de Temps', fontsize=12)
plt.ylabel('RUL (Remaining Useful Life)', fontsize=12)
plt.title('Evolution du RUL pour 5 Moteurs Différents', fontsize=14, fontweight='bold')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# Identification des capteurs constants (inutiles pour la prédiction)
print("\n🔍 Identification des capteurs constants...")
sensor_cols = [col for col in train_df.columns if col.startswith('sensor_')]
constant_sensors = []

for col in sensor_cols:
    if train_df[col].std() < 0.01:  # Variance très faible = capteur constant
        constant_sensors.append(col)

print(f"  - Capteurs constants trouvés: {constant_sensors}")
print(f"  - Nombre de capteurs constants: {len(constant_sensors)}")

# Suppression des capteurs constants
train_df_clean = train_df.drop(columns=constant_sensors)
test_df_clean = test_df.drop(columns=constant_sensors)

print(f"✓ Capteurs constants supprimés")
print(f"  - Colonnes restantes: {train_df_clean.shape[1]}")

# Analyse de corrélation
print("\n📊 Analyse de corrélation...")
remaining_sensors = [col for col in train_df_clean.columns if col.startswith('sensor_')]
corr_matrix = train_df_clean[remaining_sensors + ['RUL']].corr()

plt.figure(figsize=(14, 12))
sns.heatmap(corr_matrix, annot=True, fmt='.2f', cmap='coolwarm', center=0,
            square=True, linewidths=1, cbar_kws={"shrink": 0.8})
plt.title('Matrice de Corrélation - Capteurs vs RUL', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()

# Capteurs les plus corrélés avec RUL
rul_corr = corr_matrix['RUL'].drop('RUL').abs().sort_values(ascending=False)
print("\n🎯 Top 5 Capteurs les plus corrélés avec RUL:")
for i, (sensor, corr) in enumerate(rul_corr.head(5).items(), 1):
    print(f"  {i}. {sensor}: {corr:.3f}")

# ============================================================================
# PHASE 3: NORMALISATION DES DONNÉES
# ============================================================================

print("\n[PHASE 3] Normalisation des Données")
print("-"*70)

# Sélection des features (enlever unit_nr et time_cycles)
feature_cols = [col for col in train_df_clean.columns
                if col not in ['unit_nr', 'time_cycles', 'RUL']]

print(f"✓ Features sélectionnées: {len(feature_cols)} colonnes")

# Normalisation Min-Max [0, 1]
scaler = MinMaxScaler()
train_df_clean[feature_cols] = scaler.fit_transform(train_df_clean[feature_cols])
test_df_clean[feature_cols] = scaler.transform(test_df_clean[feature_cols])

print(f"✓ Normalisation Min-Max appliquée (intervalle [0, 1])")

# Préparation des données pour l'entraînement
X_train = train_df_clean[feature_cols]
y_train = train_df_clean['RUL']

print(f"\n📦 Dimensions des données d'entraînement:")
print(f"  - X_train: {X_train.shape}")
print(f"  - y_train: {y_train.shape}")

# ============================================================================
# PHASE 4: MODÉLISATION - MACHINE LEARNING
# ============================================================================

print("\n[PHASE 4] Modélisation et Entraînement")
print("-"*70)

# Division train/validation pour évaluation
X_train_split, X_val_split, y_train_split, y_val_split = train_test_split(
    X_train, y_train, test_size=0.2, random_state=42
)

print(f"✓ Données divisées (80% train, 20% validation)")

# ============== MODÈLE 1: RÉGRESSION LINÉAIRE (BASELINE) ==================
print("\n🤖 Modèle 1: Régression Linéaire (Baseline)")
lr_model = LinearRegression()
lr_model.fit(X_train_split, y_train_split)

# Prédictions
y_pred_lr_train = lr_model.predict(X_train_split)
y_pred_lr_val = lr_model.predict(X_val_split)

# Métriques
lr_train_rmse = np.sqrt(mean_squared_error(y_train_split, y_pred_lr_train))
lr_val_rmse = np.sqrt(mean_squared_error(y_val_split, y_pred_lr_val))
lr_val_mae = mean_absolute_error(y_val_split, y_pred_lr_val)
lr_val_r2 = r2_score(y_val_split, y_pred_lr_val)

print(f"  - RMSE (train): {lr_train_rmse:.2f} cycles")
print(f"  - RMSE (validation): {lr_val_rmse:.2f} cycles")
print(f"  - MAE (validation): {lr_val_mae:.2f} cycles")
print(f"  - R² (validation): {lr_val_r2:.3f}")

# ============== MODÈLE 2: RANDOM FOREST (AVANCÉ) ==========================
print("\n🌲 Modèle 2: Random Forest Regressor (Avancé)")
rf_model = RandomForestRegressor(
    n_estimators=100,
    max_depth=20,
    min_samples_split=5,
    random_state=42,
    n_jobs=-1
)
rf_model.fit(X_train_split, y_train_split)

# Prédictions
y_pred_rf_train = rf_model.predict(X_train_split)
y_pred_rf_val = rf_model.predict(X_val_split)

# Métriques
rf_train_rmse = np.sqrt(mean_squared_error(y_train_split, y_pred_rf_train))
rf_val_rmse = np.sqrt(mean_squared_error(y_val_split, y_pred_rf_val))
rf_val_mae = mean_absolute_error(y_val_split, y_pred_rf_val)
rf_val_r2 = r2_score(y_val_split, y_pred_rf_val)

print(f"  - RMSE (train): {rf_train_rmse:.2f} cycles")
print(f"  - RMSE (validation): {rf_val_rmse:.2f} cycles")
print(f"  - MAE (validation): {rf_val_mae:.2f} cycles")
print(f"  - R² (validation): {rf_val_r2:.3f}")

# Importance des features
feature_importance = pd.DataFrame({
    'feature': feature_cols,
    'importance': rf_model.feature_importances_
}).sort_values('importance', ascending=False)

print(f"\n🎯 Top 5 Features les plus importantes (Random Forest):")
for i, row in feature_importance.head(5).iterrows():
    print(f"  - {row['feature']}: {row['importance']:.3f}")

# ============================================================================
# PHASE 5: ÉVALUATION SUR LE JEU DE TEST
# ============================================================================

print("\n[PHASE 5] Évaluation sur le Jeu de Test")
print("-"*70)

# Prendre le dernier cycle de chaque moteur du test
test_last_cycle = test_df_clean.groupby('unit_nr').last().reset_index()
X_test = test_last_cycle[feature_cols]

print(f"✓ {len(X_test)} moteurs à évaluer (dernier cycle de chaque moteur)")

# Prédictions avec les deux modèles
y_pred_test_lr = lr_model.predict(X_test)
y_pred_test_rf = rf_model.predict(X_test)

# Comparaison avec les vraies valeurs RUL
y_test_true = rul_true['RUL'].values

# Calcul des métriques finales
test_rmse_lr = np.sqrt(mean_squared_error(y_test_true, y_pred_test_lr))
test_mae_lr = mean_absolute_error(y_test_true, y_pred_test_lr)

test_rmse_rf = np.sqrt(mean_squared_error(y_test_true, y_pred_test_rf))
test_mae_rf = mean_absolute_error(y_test_true, y_pred_test_rf)

print("\n📊 RÉSULTATS FINAUX SUR LE JEU DE TEST:")
print("\nRégression Linéaire:")
print(f"  - RMSE: {test_rmse_lr:.2f} cycles")
print(f"  - MAE:  {test_mae_lr:.2f} cycles")

print("\nRandom Forest:")
print(f"  - RMSE: {test_rmse_rf:.2f} cycles")
print(f"  - MAE:  {test_mae_rf:.2f} cycles")

# Amélioration relative
improvement = ((test_rmse_lr - test_rmse_rf) / test_rmse_lr) * 100
print(f"\n🎯 Amélioration Random Forest vs Baseline: {improvement:.1f}%")

# ============================================================================
# VISUALISATIONS FINALES
# ============================================================================

print("\n[VISUALISATIONS] Génération des graphiques...")

# Comparaison des prédictions
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Régression Linéaire
axes[0].scatter(y_test_true, y_pred_test_lr, alpha=0.6, s=100)
axes[0].plot([y_test_true.min(), y_test_true.max()],
             [y_test_true.min(), y_test_true.max()],
             'r--', linewidth=2, label='Prédiction parfaite')
axes[0].set_xlabel('RUL Réel (cycles)', fontsize=12)
axes[0].set_ylabel('RUL Prédit (cycles)', fontsize=12)
axes[0].set_title(f'Régression Linéaire\nRMSE = {test_rmse_lr:.2f}',
                  fontsize=14, fontweight='bold')
axes[0].legend()
axes[0].grid(True, alpha=0.3)

# Random Forest
axes[1].scatter(y_test_true, y_pred_test_rf, alpha=0.6, s=100, color='green')
axes[1].plot([y_test_true.min(), y_test_true.max()],
             [y_test_true.min(), y_test_true.max()],
             'r--', linewidth=2, label='Prédiction parfaite')
axes[1].set_xlabel('RUL Réel (cycles)', fontsize=12)
axes[1].set_ylabel('RUL Prédit (cycles)', fontsize=12)
axes[1].set_title(f'Random Forest\nRMSE = {test_rmse_rf:.2f}',
                  fontsize=14, fontweight='bold')
axes[1].legend()
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# Distribution des erreurs
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

errors_lr = y_test_true - y_pred_test_lr
errors_rf = y_test_true - y_pred_test_rf

axes[0].hist(errors_lr, bins=20, alpha=0.7, edgecolor='black')
axes[0].axvline(0, color='red', linestyle='--', linewidth=2)
axes[0].set_xlabel('Erreur de Prédiction (cycles)', fontsize=12)
axes[0].set_ylabel('Fréquence', fontsize=12)
axes[0].set_title('Distribution des Erreurs - Régression Linéaire',
                  fontsize=14, fontweight='bold')
axes[0].grid(True, alpha=0.3)

axes[1].hist(errors_rf, bins=20, alpha=0.7, color='green', edgecolor='black')
axes[1].axvline(0, color='red', linestyle='--', linewidth=2)
axes[1].set_xlabel('Erreur de Prédiction (cycles)', fontsize=12)
axes[1].set_ylabel('Fréquence', fontsize=12)
axes[1].set_title('Distribution des Erreurs - Random Forest',
                  fontsize=14, fontweight='bold')
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# ============================================================================
# TABLEAU COMPARATIF FINAL
# ============================================================================

print("\n" + "="*70)
print("TABLEAU COMPARATIF DES PERFORMANCES")
print("="*70)

comparison_df = pd.DataFrame({
    'Modèle': ['Régression Linéaire', 'Random Forest'],
    'RMSE (Validation)': [lr_val_rmse, rf_val_rmse],
    'RMSE (Test)': [test_rmse_lr, test_rmse_rf],
    'MAE (Test)': [test_mae_lr, test_mae_rf],
    'R² (Validation)': [lr_val_r2, rf_val_r2]
})

print(comparison_df.to_string(index=False))

print("\n" + "="*70)
print("CONCLUSION ET RECOMMANDATIONS")
print("="*70)
print(f"""
✓ Le modèle Random Forest présente les meilleures performances
✓ RMSE sur le test: {test_rmse_rf:.2f} cycles
✓ Amélioration de {improvement:.1f}% par rapport au baseline

⚠ FIABILITÉ POUR UN USAGE RÉEL EN AÉRONAUTIQUE:
  - Une erreur de {test_rmse_rf:.0f} cycles peut être acceptable selon le contexte
  - Pour une sécurité optimale, ajouter une marge de sécurité de 20-30%
  - Recommandation: Combiner avec des inspections visuelles
  - Nécessité de validation sur des données réelles d'avions
  - Mise en place d'un système d'alertes multicouches

📋 PROCHAINES ÉTAPES:
  1. Tester avec des architectures Deep Learning (LSTM, CNN)
  2. Intégrer plus de données de vol réelles
  3. Développer un système d'alerte avec seuils adaptatifs
  4. Validation croisée avec experts aéronautiques
""")

print("\n✅ Analyse terminée avec succès!")
```
